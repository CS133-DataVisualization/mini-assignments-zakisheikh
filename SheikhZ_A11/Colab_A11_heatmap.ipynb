```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

IGN="https://raw.githubusercontent.com/csbfx/advpy122-data/master/ign_scores.csv"
## Your code here . . .
# Load the IGN data into a DataFrame
ign_scores = pd.read_csv(IGN)

# Set the "Platform" column as the index for better labeling in visualizations
ign_scores.set_index("Platform", inplace=True)

# Preview the data structure
ign_scores.head()
```

#### Problem 4 - (2 points)
Since your instructor's gaming interests are pretty broad, you can help her decide to use the IGN scores to determine the choice of genre and platform.

1. Create a heatmap using the IGN scores by genre and platform and include the scores in the cells of the heatmap.
2. Base on the heatmap, which combination of genre and platform receives the highest average ratings?  Which combination receives the lowest average rankings? Write the answers in a markdown cell.

```python
## Create a heatmap
# Your code here . . .
# Create a heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    ign_scores,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f",
    linewidths=0.5,
    cbar_kws={"label": "IGN Score"},
)
plt.title("IGN Scores Heatmap by Genre and Platform")
plt.xlabel("Genre")
plt.ylabel("Platform")
plt.tight_layout()
plt.show()

# Finding highest and lowest average combinations
highest_avg = ign_scores.stack().idxmax()
lowest_avg = ign_scores.stack().idxmin()

highest_avg, lowest_avg
```

## Problem 5 - (4 points)
Use the Pokemon dataset to create a clustermap with color. First, filter the dataframe to only keep data with `Type 1` equals to following values: `Water`, `Normal`, `Grass`, `Bug` and `Psychic`. Annotate the dendrogram using different colors for these five different `Type 1` values. Use `Name` as the index.

pokemon_data is in https://raw.githubusercontent.com/csbfx/advpy122-data/master/Pokemon.csv

```python
# Problem 5
# Your code here . . .
pkdf = "https://raw.githubusercontent.com/csbfx/advpy122-data/master/Pokemon.csv"
pokemon_df= pd.read_csv(pkdf)

# Filter for specified Type 1 values
filtered_pokemon = pokemon_df[
    pokemon_df["Type 1"].isin(["Water", "Normal", "Grass", "Bug", "Psychic"])
]

# Use Name as index
filtered_pokemon.set_index("Name", inplace=True)

# Drop non-numerical columns for clustering
clustering_data = filtered_pokemon.drop(
    columns=["#", "Type 1", "Type 2", "Generation", "Legendary"]
)

# Create clustermap
sns.clustermap(
    clustering_data,
    cmap="coolwarm",
    annot=False,
    figsize=(12, 8),
    col_cluster=False,
    row_colors=filtered_pokemon["Type 1"].map(
        {
            "Water": "blue",
            "Normal": "gray",
            "Grass": "green",
            "Bug": "orange",
            "Psychic": "purple",
        }
    ),
)
plt.title("Pokemon Clustermap by Type")
plt.show()
```
