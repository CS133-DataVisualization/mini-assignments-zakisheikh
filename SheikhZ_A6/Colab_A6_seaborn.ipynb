Here are the descriptions of this dataset:
- `#` - Pokedex entry number of the Pokemon
- `Name` - name of the Pokemon
- `Type 1` - each Pokemon has a type, this determines weakness/resistance to attacks [referred to as the primary type]
- `Type 2` - some Pokemon are dual type and have 2 [referred to as the secondary type]
- `Total` - sum of all stats that come after this, a general guide to how strong a Pokemon is
- `HP` - hit points, or health, defines how much damage a Pokemon can withstand before fainting
- `Attack` - the base modifier for normal attacks
- `Defense` - the base damage resistance against normal attacks
- `Sp. Atk` - special attack, the base modifier for special attacks
- `Sp. Def` - the base damage resistance against special attacks
- `Speed` - determines which Pokemon attacks first each round
- `Generation` - refers to which grouping/game series the Pokemon was released in
- `Legendary` - a boolean that identifies whether the Pokemon is legendary

## Q1: Plot the relationship between Attack and Defense
Create a relationship plot using seaborn's `replot` function. Plot `Attack` on the x-axis and `Defense` on the y-axis. Use `hue` to display the `HP` (hit point) for each datapoint. You can choose your favorite palette. Include a meaninful title in the plot. After creating the plot, insert a markdown cell immediately following the plot to describe the relationship between Attack and Defense in this dataset.

```python
# Your code here . . .
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Mocking data generation based on dataset descriptors (since we don't have the CSV)
# Generate random data within reasonable ranges for the columns
url = "https://raw.githubusercontent.com/csbfx/advpy122-data/master/Pokemon.csv"
df = pd.read_csv(url)

# Plotting the relationship between Attack and Defense
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Attack', y='Defense', hue='HP', palette='coolwarm')
plt.title("Relationship between Attack and Defense (Hue: HP)")
plt.xlabel("Attack")
plt.ylabel("Defense")
plt.grid(True)
plt.show()
```

From the plot, I observed....


* Pokémon with varying HP values are distributed across different combinations of Attack and Defense, showing diversity in their attributes.
* No clear linear trend exists between Attack and Defense, suggesting independent design of these stats.

## Q2: Plot the relationship between Attack and Total using a hexbin plot
Create a hexbin plot using seaborn's `joinplot` function. Plot `Attack` on the x-axis and `Total` on the y-axis. You can choose the gridsize. Include a title for the plot "Attack vs total for all Pokemon". After generating the plot, insert a markdown cell to answer this question: The hexbin plot displays the relationship between `Attack` and `Total`, what else does this plot show?

```python
# Your code here . . .
# Plot the hexbin plot for Attack vs Total
plt.figure(figsize=(10, 6))
hexbin_plot = sns.jointplot(data=df, x='Attack', y='Total', kind='hex', gridsize=30, cmap='Blues')
hexbin_plot.ax_joint.set_title("Attack vs Total for all Pokémon", fontsize=14, pad=20)
hexbin_plot.ax_joint.set_xlabel("Attack", fontsize=12)
hexbin_plot.ax_joint.set_ylabel("Total", fontsize=12)
plt.show()

```

From the plot, I observed....


* Pokémon with higher attack stats generally have higher total stats, showing a positive correlation.
* Most Pokémon fall in the attack range of 50-100 and total range of 200-400, representing the majority of non-legendary or mid-tier Pokémon.
* Pokémon with attack stats above 150 and total stats above 500 are less common, likely representing legendary Pokémon or final-stage evolutions.




## Q3: Create two categorical plots for Pokemon types. `Type 1` and `Type 2` .
For each plot, the unique values of `Type 1` or `Type 2` should be plotted along the y-axis, and the counts of Pokemon should be on the x-axis. Use the color red for the `Type 1` plot and blue for the `Type 2` plot. After the two plots are displayed, insert a markdown cell to write your findings: Which are the most common `Type 1` and `Type 2` Pokemons, and which are the most rare ones?

```python
# Your code here . . .
# Create the Type 1 categorical plot
plt.figure(figsize=(10, 8))
sns.countplot(y='Type 1', data=df, color='red', order=df['Type 1'].value_counts().index)
plt.title("Count of Pokémon by Type 1", fontsize=14)
plt.xlabel("Count", fontsize=12)
plt.ylabel("Type 1", fontsize=12)
plt.show()

# Create the Type 2 categorical plot
plt.figure(figsize=(10, 8))
sns.countplot(y='Type 2', data=df, color='blue', order=df['Type 2'].value_counts().index)
plt.title("Count of Pokémon by Type 2", fontsize=14)
plt.xlabel("Count", fontsize=12)
plt.ylabel("Type 2", fontsize=12)
plt.show()
```

From the plot, I observed....

#### Type 1
* The most common Type 1 Pokémon are Water and Normal, likely because these types include many early-stage or common Pokémon.
* The rarest Type 1 Pokémon are Fairy and Flying, which tend to have fewer representatives and often unique characteristics in the Pokémon universe.

#### Type 2:
* The most common Type 2 Pokémon are Flying and Ground, as many Pokémon have these as secondary types (e.g., Bug/Flying, Grass/Poison).
* The rarest Type 2 Pokémon are Normal and Bug, reflecting their specialized nature and limited availability.
